```python
#dictionary of bigrams
b = {}
for w in words:
  # starting and ending special value characters
  chs = ['<S>'] + list(w) + ['<E>']
  # iterates starting from the first and second characters in the word
  # ends when we can't collect ch1 and ch2
  for ch1, ch2 in zip(chs, chs[1:]):
    bigram = (ch1,ch2)
    # adds to the bigram if doesn't exist, and tracks frequency
    b[bigram] = b.get(bigram, 0) + 1
```

```python
# sorting b.items() by getting the second val of each tuple and descending
sorted(b.items(), key = lambda kv: -kv[1])
```
