```python
!pip install transformers torchvision pillow

# Import necessary modules
from transformers import CLIPProcessor, CLIPModel
from PIL import Image as PILImage
import torch
import numpy as np

# Load the CLIP model and processor
modelArray = ["openai/clip-vit-base-patch16", "openai/clip-vit-base-patch32"]

def image_embedding(image_path, modelNum):
    """
    Takes an image file path and returns its embedding as a NumPy array.
    """
    print("Using model " + modelArray[modelNum])
    model = CLIPModel.from_pretrained(modelArray[modelNum])
    processor = CLIPProcessor.from_pretrained(modelArray[modelNum])

    # Load and preprocess the image
    image = PILImage.open(image_path).convert("RGB")
    inputs = processor(images=image, return_tensors="pt")

    # Extract image features
    with torch.no_grad():
        embeddings = model.get_image_features(**inputs)

    # Normalize and convert to NumPy
    embeddings = embeddings / embeddings.norm(p=2, dim=-1, keepdim=True)
    return embeddings.squeeze().cpu().numpy()
```

```python
def yVector(emotion):
    if emotion == "happy":
        return [0, 0, 1, 0, 0]
    elif emotion == "sadness":
        return [0, 0, 0, 1, 0]
    elif emotion == "anger":
        return [1, 0, 0, 0, 0]
    elif emotion == "fear":
        return [0, 1, 0, 0 ,0]
    elif emotion == "surprise":
        return [0, 0, 0, 0, 1]
    return None
```

```python
import random
def create_embedding_matrix_function (wordlist,modelNum,embeddings_as_rows = True):
  #image_files #an array containing name of all image file in the current folder.
  xArray = []
  yArray = []
  for name in wordlist:
    path = name
    print("x matrix added row embedded: " + path)
    vector = image_embedding(path,modelNum)
    xArray.append(vector)

    emotion = path.partition(" ")[0]
    yArray.append(yVector(emotion))
    print("Y matrix append new emotion: " + emotion)
    print(str(len(xArray)) + "\n")
  if embeddings_as_rows == True:
    x_embedding_matrix = np.stack(xArray, axis = 0)
    y_embedding_matrix = np.stack(yArray, axis = 0)
  else:
    x_embedding_matrix = np.stack(xArray, axis = 1)
    y_embedding_matrix = np.stack(yArray, axis = 1)

  np.save("embeddings_x.npy",x_embedding_matrix)
  np.save("embeddings_y.npy",y_embedding_matrix)
  return x_embedding_matrix,y_embedding_matrix
```

```python
def model_fit_pytorch(x_train, y_train, epochs, lr, model, log_interval,loss_criterion = nn.MSELoss()):

    optimizer = optim.SGD(model.parameters(), lr = lr) # SGD is (Stochastic) Gradient Descent

    loss_history = []

    for epoch in range(epochs):

        model.train() # enter training mode
        optimizer.zero_grad() # reset the gradients of all model parameters before computing new gradients

        # predict output based on the model
        y_pred = model(x_train)

        # Compute loss
        loss = loss_criterion(y_pred, y_train)

        loss.backward()
        optimizer.step()

        loss_history.append(loss.item())

        if log_interval and epoch % log_interval == 0:
            print(f"Epoch [{epoch}/{epochs}], Loss: {loss_history[-1]:.4f}")

    return model, loss_history
```

```python
import numpy as np

def compute_confusion_matrix(correct_labels, predicted_labels):
    unique_labels = np.unique(correct_labels)
    num_labels = len(unique_labels)

    confusion_matrix = np.zeros((num_labels, num_labels))

    for i, true_label in enumerate(unique_labels):
        true_idxs = (correct_labels == true_label)
        num_true = np.sum(true_idxs)

        for j, pred_label in enumerate(unique_labels):
            num_pred_as_label = np.sum(predicted_labels[true_idxs] == pred_label)
            if num_true > 0.0:
                confusion_matrix[i, j] = num_pred_as_label/num_true

    return confusion_matrix, unique_labels
```

```python
import numpy as np
import seaborn as sns
import torch
import matplotlib.pyplot as plt
import torch.nn as nn

# --- Compute Confusion Matrix ---
def compute_confusion_matrix(correct_labels, predicted_labels):
    correct_labels = np.array(correct_labels)
    predicted_labels = np.array(predicted_labels)

    unique_labels = np.unique(correct_labels)
    num_labels = len(unique_labels)
    confusion_matrix = np.zeros((num_labels, num_labels))

    for i, true_label in enumerate(unique_labels):
        true_idxs = (correct_labels == true_label)
        num_true = np.sum(true_idxs)

        for j, pred_label in enumerate(unique_labels):
            num_pred_as_label = np.sum(predicted_labels[true_idxs] == pred_label)
            if num_true > 0:
                confusion_matrix[i, j] = num_pred_as_label / num_true

    return confusion_matrix, unique_labels

# --- Plot Heatmap ---
def plot_confusion_matrix_heatmap(conf_matrix, class_labels=None, title="Confusion Matrix", test_loss=None):
    sns.heatmap(conf_matrix, annot=True, fmt=".2f", cmap="cividis",
                xticklabels=class_labels, yticklabels=class_labels)
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    if test_loss is not None:
        plt.title(f"{title}\nTest Loss: {test_loss:.4f}")
    else:
        plt.title(title)

# --- Function to evaluate test loss ---
def evaluate_test_loss(model, x_test_tensor, y_test_tensor, criterion):
    model.eval()
    with torch.no_grad():
        y_pred = model(x_test_tensor)
        loss = criterion(y_pred, y_test_tensor)
    return loss.item()

# --- Define loss function (adjust based on your targets) ---
criterion = nn.MSELoss()

# --- Emotion labels ---
emotion_labels = ["anger", "fear", "happy", "sadness", "surprise"]

# --- Prepare data for Model 1 ---
m1_trained_multi_class_model.eval()
with torch.no_grad():
    m1_preds = torch.argmax(m1_trained_multi_class_model(m1_x_test_matrix_tensor), dim=1).numpy()
m1_true = torch.argmax(m1_y_test_matrix_tensor, dim=1).numpy()
m1_test_loss = evaluate_test_loss(m1_trained_multi_class_model, m1_x_test_matrix_tensor, m1_y_test_matrix_tensor, criterion)

# --- Prepare data for Model 2 ---
m2_trained_multi_class_model.eval()
with torch.no_grad():
    m2_preds = torch.argmax(m2_trained_multi_class_model(m2_x_test_matrix_tensor), dim=1).numpy()
m2_true = torch.argmax(m2_y_test_matrix_tensor, dim=1).numpy()
m2_test_loss = evaluate_test_loss(m2_trained_multi_class_model, m2_x_test_matrix_tensor, m2_y_test_matrix_tensor, criterion)

# --- Compute both confusion matrices ---
m1_conf, _ = compute_confusion_matrix(m1_true, m1_preds)
m2_conf, _ = compute_confusion_matrix(m2_true, m2_preds)

# --- Plot side-by-side ---
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plot_confusion_matrix_heatmap(m1_conf, class_labels=emotion_labels, title="Model 1 Confusion Matrix", test_loss=m1_test_loss)

plt.subplot(1, 2, 2)
plot_confusion_matrix_heatmap(m2_conf, class_labels=emotion_labels, title="Model 2 Confusion Matrix", test_loss=m2_test_loss)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

def rank_rows_by_similarity_matrix(cosine_similarity_matrix):

    n = cosine_similarity_matrix.shape[0]

    # Copy so we don't overwrite the inpu
    sums = np.sum(cosine_similarity_matrix, axis=1)

    sorted_row_indices = np.argsort(-sums)


    for i in range(n):
        rank_matrix = cosine_similarity_matrix[sorted_row_indices, :]
        rank_matrix = rank_matrix[:, sorted_row_indices]
    return rank_matrix
```
